# DFT von Signalen in Abhängigkeit von Abtastfrequenz und Signallänge
Zeichnet ein Signal mit einstellbarer Signalfrequenz und Abtastrate und berechnet per DFT das Spektrum. Das Zeitfenster für die DFT kann verändert werden.

- Verändern Sie die Signalfrequenz inkrementell, und verwenden Sie verschiedene Fensterfunktionen. Wie verändert sich jeweils die Maximalamplitude der Peaks mit der Frequenz?

Hinweis: Wenn Sie einen Slider anclicken, können Sie die Pfeiltasten (je nach System + "Strg"-Taste) zum Verschieben benutzen

```python
"""
DFT-Demoscript (Matlab-Style)

(c) Benno Dömer | HS Pforzheim 2022-2026

"""
import platform
if platform.system() == "Emscripten": # running in JupyterLite
    %pip install -r runtime-requirements.txt

%matplotlib widget 

from numpy import (pi, log10, exp, sqrt, sin, cos, tan, angle, arange, linspace, zeros, ones, hamming, hanning, blackman) # numpy für's Rechnen, bei Bedarf anpassen
from numpy.fft import fft
from matplotlib.pyplot import (figure, plot, stem, legend, text, show,
                               grid, xlabel, ylabel, subplot, title, clf, cla, xlim, ylim) # matpotlib für's  Plotten, bei Bedarf anpassen
from matplotlib.pyplot import fill_between as area # Beispiel für eine Mapping auf Matlab-Bezeichnung
import ipywidgets as widgets # interaktive Controls wie Slider, Dropdowns etc.

fig = figure(figsize=(9, 5), dpi=100, layout='constrained',num=' ') 

def myplotstyle(ax): # Hilfsfunktion, hübscht den Plot etwas auf
    ax.spines['right'].set_color('none')
    ax.spines['top'].set_color('none')
    ax.xaxis.set_ticks_position('bottom')
    ax.spines['bottom'].set_position(('data',0))
    ax.yaxis.set_ticks_position('left')
    ax.spines['left'].set_position(('data',0))
 

   
def plotdata(fa,tl,f0,xax,wd):
    #figure(fig)
    
    ax1=subplot(3,1,1)
    cla()
    nw = round(fa * tl);
    nwhr = round(tl*100);
    print(f"Fensterlänge: {nw*1/fa:.1f} s, N= {nw} Werte")
    tc = arange(-2,10,0.01); # Zeitachste für Darstellung "kontiniuierliches" Signal
    tw = tc[0:nwhr]
    yc = cos(2*pi*f0*tc); # "kontiniuierliches" Signal 
    
    t = arange(0,10,1/fa); # Abtastzeitpunkte
    y = cos(2*pi*f0*t); # abgetastetes Signal

    
    if wd == "Hamming": # Hamming
        fwd = hamming(nw); # Hammingfenster
        yw = hamming(nwhr)
    elif wd == "Hanning": # Hanning
        fwd = hanning(nw); # Hanningfenster
        yw = hanning(nwhr)
    elif wd == "Blackman": # Blackman
        fwd = blackman(nw); # Blackmanfenster
        yw = blackman(nwhr)
    else: # Default: Rechteck
        fwd = ones(nw); # Rechteckfenster
        yw = ones(nwhr)
    

    
    plot(tc,yc,#  Kontinuierliches Signal
         t[0:nw],y[0:nw],'.-r', # Abtastpunkte im Signalausschnitt,
         tw+2,yw,'-k',
         [0, 0, nw*1/fa, nw*1/fa],[0, 1, 1, 0],':k') # Fenster
    xlabel('t / s',loc='right')
    ylabel('y(t)',loc='top')
    title(r'y(t) = $cos(2\pi f_0 t)$ mit $f_0$ = ' f'{f0:.1f} Hz. Rot: Abtastpunkte mit $f_A$ = {fa:.1f}  Hz')
    ylim(-1.5,1.5)
    myplotstyle(ax1)

    ax2=subplot(3,1,2)
    cla()
    #stem(arange(0,nw,1),y[0:nw],'r') # y[k]; X-Achse um 1 korrigiert wg. Zählung ab k=0
    stem(arange(0,nw,1),y[0:nw], linefmt=':k', markerfmt='k.', basefmt="k-") # y[k]; X-Achse um 1 korrigiert wg. Zählung ab k=0
    # linefmt=':k', markerfmt='ko', basefmt="k-"
    y=y[0:nw]*fwd
    stem(arange(0,nw,1),y[0:nw], linefmt='-r', markerfmt='ro', basefmt="k-") # y[k]; X-Achse um 1 korrigiert wg. Zählung ab k=0
    xlabel('k',loc='right')
    ylabel('y[k]')
    title(r'y[k] = $cos(2\pi f_0 k\cdot T_A)$ mit $T_A = 1/f_A =$' +f' {1/fa:.3f} s, rot: mit Fensterfunktion')
    ylim(-1.3,1.3)
    myplotstyle(ax2)
    
    ax3=subplot(3,1,3)
    cla()
    Y = fft(y,nw);
    if xax:
        stem(arange(0,nw)*fa/nw,abs(Y)) # Y[k]; X-Achse um 1 korrigiert wg. Zählung ab n=0    
        plot([fa/2,fa/2],[0, 3*nw/4],':r')
        ax3.text(f0+.1, nw/4, f'$f_0$ = {f0:.2f} Hz')
        ax3.text(fa/2+.1, nw/4,f'$f_A/2$ = {fa/2:.1f} Hz')
        xlabel('f / Hz',loc='right')
    else:
        stem(arange(0,nw),abs(Y)) # Y[k]; X-Achse um 1 korrigiert wg. Zählung ab n=0    
        xlabel('n',loc='right')
    ylabel('|Y[n]|')
    myplotstyle(ax3)

    title(f'DFT-Spektrum von y[t]. Fensterlänge: {nw/fa:.1f} s, Auflösung $\\Delta f= ${fa/nw:.3f} Hz')
    #set(gcf,'Position',[0,0,700,150])
    fig.canvas.draw()
    #show()
    
## Interaktion: 
# Definition von Controls: 
faslider  = widgets.FloatSlider(min=1, max=50, step=1, value=10, continuous_update=False, description='$f_A$')
tlslider  = widgets.FloatSlider(min=1, max=10, step=0.1, value=5, continuous_update=False, description='$t_l$')
f0slider  = widgets.FloatSlider(min=-1, max=10, step=.05, value=1, continuous_update=True, description='$f_0$')
axcombo = widgets.Dropdown(options=[('Frequenz',True),('Stützstellen (n)',False)],
                                    value=True,description='x-Achse')
sigcombo = widgets.Dropdown(options=['Rechteck','Hamming','Hanning','Blackman'],
                                    value='Rechteck',description='Fenster:')


# Arrangieren hier in horizontale Box:
ui1 = widgets.HBox([f0slider, tlslider, faslider])
ui2 = widgets.HBox([sigcombo, axcombo])
                                    

#Verknüpfen mit Callback
out = widgets.interactive_output(plotdata, { 'fa': faslider,'tl': tlslider,'f0':f0slider,'xax':axcombo, 'wd':sigcombo}) 
#Anzeigen
display(ui1,ui2,out)
show()
```
